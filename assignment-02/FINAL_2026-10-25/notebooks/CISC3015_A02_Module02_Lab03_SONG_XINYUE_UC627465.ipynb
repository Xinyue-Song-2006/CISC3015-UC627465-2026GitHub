```python
import numpy as np
import matplotlib.pyplot as plt
```

## Line, bar, scatter, and area plots
These plots compare a small reproducible dataset and show how chart choice changes the story.

```python
years = np.arange(2018, 2024)
visitors = np.array([120, 155, 180, 205, 240, 275])
revenue = np.array([42, 48, 55, 63, 71, 84])
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0, 0].plot(years, visitors, marker='o', color='steelblue')
axes[0, 0].set_title('Visitor trend')
axes[0, 1].bar(years, revenue, color='darkorange')
axes[0, 1].set_title('Revenue by year')
axes[1, 0].scatter(visitors, revenue, s=90, color='seagreen')
axes[1, 0].set_xlabel('Visitors')
axes[1, 0].set_ylabel('Revenue')
axes[1, 0].set_title('Visitors vs revenue')
axes[1, 1].fill_between(years, visitors, alpha=0.35, color='mediumpurple')
axes[1, 1].set_title('Cumulative area view')
for ax in axes.flat:
    ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```
